```python
seek_codes = pd.read_csv('/archive/vision/beery/animal_reid/datasets/elephants_zooniverse/ele_id_project/data/out_apr2/SEEK_codes_.csv')

# Initialize lists to collect data for the expanded DataFrame
elephant_ids = []
characteristics = []
total_votes_list = []
ground_truth_values = []
correct_vote_counts = []
wrong_vote_counts = []
unknown_vote_counts = []

# Process each elephant in seek_df
for elephant, characteristics_dict in seek_dict.items():
    for characteristic, votes in characteristics_dict.items():
        total_votes = sum(votes.values())
        
        # Identify ground truth for this characteristic from SEEK_codes
        ground_truth = seek_codes.loc[seek_codes['ele_id'] == elephant, characteristic].values
        if ground_truth.size == 0:
            continue  # Skip if no ground truth exists
        
        ground_truth = ground_truth[0]  # Ground truth value
        correct_votes = votes.get(str(ground_truth), 0) if ground_truth != '_' else 0
        unknown_votes = votes.get('_', 0)
        
        # Calculate wrong votes as those not equal to ground truth and not unknown
        wrong_votes = sum(count for vote, count in votes.items() if vote != str(ground_truth) and vote != '_')
        
        # Collect data for the current characteristic
        elephant_ids.append(elephant)
        characteristics.append(characteristic)
        total_votes_list.append(total_votes)
        ground_truth_values.append(ground_truth)
        correct_vote_counts.append(correct_votes)
        wrong_vote_counts.append(wrong_votes)
        unknown_vote_counts.append(unknown_votes)

# Build the expanded DataFrame with counts
result_df = pd.DataFrame({
    'Elephant_ID': elephant_ids,
    'Characteristic': characteristics,
    'Total_Votes': total_votes_list,
    'Accepted_SEEK_code': ground_truth_values,
    'Agree': correct_vote_counts,
    'Disagree': wrong_vote_counts,
    'Unknown': unknown_vote_counts
})

# Calculate proportions
result_df['Proportion_Agree'] = result_df['Agree'] / result_df['Total_Votes']
result_df['Proportion_Disagree'] = result_df['Disagree'] / result_df['Total_Votes']
result_df['Proportion_Unknown'] = result_df['Unknown'] / result_df['Total_Votes']

# Display the final DataFrame
result_df
```

```python
# Group by 'Characteristic' and calculate the mean for numeric columns only
aggregated_df = result_df.groupby('Characteristic').mean(numeric_only=True).reset_index()

# Display the aggregated DataFrame
aggregated_df
```

```python
# Assuming 'aggregated_df' is the DataFrame with the averaged values for each characteristic
num_columns = 6
num_rows = (len(aggregated_df) + num_columns - 1) // num_columns

fig, axes = plt.subplots(num_rows, num_columns, figsize=(20, 10))
axes = axes.flatten()

# Loop over each characteristic to plot the aggregated data
for i, (index, row) in enumerate(aggregated_df.iterrows()):
    data = row[['Agree', 'Disagree', 'Unknown']]
    axes[i].bar(data.index, data.values)
    axes[i].set_title(f"{row['Characteristic']}")

# Hide any unused subplots
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

```python
import sys
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from collections import Counter
from seek_code import SEEK

# Add custom path
sys.path.append('/data/vision/beery/scratch/antoine/CBM_reid/methods')

# Load data
image_dictonary = pd.read_csv('/data/vision/beery/scratch/antoine/CBM_reid/data_processing/data/out_apr2/image_dictonary.csv')
subject_SEEK = image_dictonary['subject-SEEK']

# Prepare data for plotting
data = []
for characteristic in SEEK.attribute_names:
    attribute_values = [getattr(SEEK(subject_SEEK[i]), characteristic) for i in range(len(subject_SEEK))]
    attribute_counts = Counter(attribute_values)
    for value, count in attribute_counts.items():
        data.append({'Attribute': characteristic.capitalize(), 'Value': value, 'Count': count})

# Convert to DataFrame for Seaborn
plot_data = pd.DataFrame(data)

# Initialize the grid layout
num_attributes = len(SEEK.attribute_names)
cols = 4  # Fit more plots per row
rows = (num_attributes // cols) + (1 if num_attributes % cols != 0 else 0)

# Plot each attribute on a tighter grid
fig, axes = plt.subplots(rows, cols, figsize=(16, rows * 3))
axes = axes.flatten()

sns.color_palette("magma")

for i, characteristic in enumerate(plot_data["Attribute"].unique()):
    ax = axes[i]
    subset = plot_data[plot_data["Attribute"] == characteristic]
    

    sns.barplot(
        x="Value", 
        y="Count", 
        data=subset, 
        ax=ax, 
        palette="pastel"
    )
    
    # Calculate the majority ratio
    total_count = subset["Count"].sum()
    max_count = subset["Count"].max()
    majority_ratio = max_count / total_count
    
    # Add the ratio to the plot
    ax.text(
        0.95, 
        0.95, 
        f"Ratio: {majority_ratio:.2f}", 
        transform=ax.transAxes, 
        fontsize=10, 
        color="black", 
        ha="right", 
        va="top", 
        bbox=dict(boxstyle="round,pad=0.3", edgecolor="gray", facecolor="white", alpha=0.8)
    )
    
    # Customize plot
    ax.set_title(characteristic, fontsize=10, fontweight="bold", pad=5)
    ax.set_xlabel("")
    ax.set_ylabel("")
    ax.tick_params(axis="x", labelsize=8, rotation=45)
    ax.tick_params(axis="y", labelsize=8)

# Remove any unused subplots
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

# Tight layout to minimize spacing
plt.subplots_adjust(wspace=0.4, hspace=0.6)  # Adjust width and height spacing
plt.show()
```
